```python
x=df.iloc[:,15:20].values
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_scaled=sc.fit_transform(x)
```

```python
import matplotlib.pyplot as plt
import numpy as np
# import seaborn as sns
from sklearn.decomposition import PCA
pca=PCA()
x_pca=pca.fit_transform(x_scaled)
#Scree Plot to find out optimal no.of components that explains maximum variance

plt.figure(figsize=(10,5))
plt.plot(range(1,len(pca.explained_variance_ratio_)+1),pca.explained_variance_ratio_,'o-',label='Individual Components Variance Ratio')
# o- => marker='o' and linestyle='-'
plt.plot(range(1,len(pca.explained_variance_ratio_)+1),np.cumsum(pca.explained_variance_ratio_),'s--',label='Cumulative Components Variance Ratio')
plt.xlabel('No.of PCA Components')
plt.ylabel('Explained Variance')
plt.legend()
plt.grid(True)
plt.show()
print(pca.explained_variance_)
print(pca.explained_variance_ratio_)
```

```python
pca=PCA(n_components=4)
x_reduced=pca.fit_transform(x_scaled)
print(x_scaled.shape)
print(x_reduced.shape)
```

```python
plt.scatter(x_reduced[:,0],x_reduced[:,1])
plt.xlabel('PCA1')
plt.ylabel('PCA2')
plt.show()
```

```python
from sklearn.cluster._kmeans import KMeans
from sklearn.metrics import silhouette_score
# wcss=[]
for k in range(2,10):
    kmc=KMeans(n_clusters=k,init='k-means++',random_state=42)
    clusters=kmc.fit_predict(x_reduced)
    score=silhouette_score(x_reduced,clusters)
    print(f'k={k} silhouette score={score}')
```

```python
kmc=KMeans(n_clusters=3,init='k-means++')
clusters=kmc.fit_predict(x_reduced)
plt.scatter(x[:,0],x[:,3],c=kmc.labels_,cmap='viridis',label='data points')
plt.scatter(kmc.cluster_centers_[:,0],kmc.cluster_centers_[:,1],marker='X',label='Centroids')
plt.xlabel('NumDealsPurchases')
plt.ylabel('NumwebPurchases')
plt.title('Clusters')
plt.legend()
plt.show()
```
