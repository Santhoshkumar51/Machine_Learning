```python
import numpy as np 
x=np.array([1,2,3,4,5,6]).reshape(-1,1)
y=np.array([35,40,50,60,65,70])
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lg=LinearRegression()
lg.fit(x_train,y_train)
# ytrain_pred=lg.predict(x_train)
ytest_pred=lg.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error
print("r2 score: ",r2_score(y_test,ytest_pred))
print("mse: ",mean_squared_error(y_test,ytest_pred))
```

```python
from sklearn.preprocessing import PolynomialFeatures
pf=PolynomialFeatures(degree=6)
x_train_poly=pf.fit_transform(x_train)
x_test_poly=pf.transform(x_test)
```

```python
lg2=LinearRegression()
lg2.fit(x_train_poly,y_train)
y_pred=lg2.predict(x_test_poly)
print("r2 score: ",r2_score(y_pred,y_test))
print("mse: ",mean_squared_error(y_test,y_pred))
# print(lg2.score(y_test,x_test_poly))
```
